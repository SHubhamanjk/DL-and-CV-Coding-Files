```python
import cv2
```

Reading and Visualizing an Image

```python
image_path="happy new year.png"
image = cv2.imread(image_path)
cv2.imshow("image",image)
cv2.waitKey(0)
```

Reading and Visualizing a Video

```python
video_path="VID-20241130-WA0088.mp4"
video = cv2.VideoCapture(video_path)
ret=True
while ret:
    ret,frame= video.read()
    if ret:
        cv2.imshow("video",frame)
        cv2.waitKey(30)

video.release()
cv2.destroyAllWindows()        
```

Reading and Visualizing from Webcam

```python
webcam=cv2.VideoCapture(0)
ret=True
while ret:
    ret,frame= webcam.read()
    if ret:
        cv2.imshow("webcam",frame)
        if cv2.waitKey(30) & 0xFF==ord("q"):
            break
webcam.release()
cv2.destroyAllWindows()        
```

ColorSpaces

```python
image_path="mypic.jpg"
rgb_image = cv2.imread(image_path)
rgb_image=cv2.resize(rgb_image,(400,650))
cv2.imshow("rgb_image",rgb_image)
cv2.imshow("bgr_image",cv2.cvtColor(rgb_image,cv2.COLOR_BGR2RGB))
cv2.imshow("gray_image",cv2.cvtColor(rgb_image,cv2.COLOR_BGR2GRAY))
cv2.imshow("hsv_image",cv2.cvtColor(rgb_image,cv2.COLOR_BGR2HSV))
cv2.imshow("hls_image",cv2.cvtColor(rgb_image,cv2.COLOR_BGR2HLS_FULL))
cv2.imshow("luv_image",cv2.cvtColor(rgb_image,cv2.COLOR_BGR2LUV))


cv2.waitKey(0)
```

Also there is concept of 
Threshold : used to remove unwant noise from image although blur also use for the same
Edge : used to draw edge of an image
also we can draw multiple kind of drawings in the image

```python
image = cv2.imread('mypic.jpg')

gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

_, thresholded = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)

cv2.imshow("Original Image", image)
cv2.imshow("Thresholded Image", thresholded)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
image = cv2.imread('mypic.jpg')

cv2.rectangle(image, (50, 50), (200, 200), (0, 255, 0), 3)

cv2.circle(image, (300, 300), 50, (0, 0, 255), 3)

cv2.line(image, (400, 100), (500, 200), (255, 0, 0), 2)

cv2.putText(image, 'SHUBHAM', (150, 400), cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 255, 0), 2, cv2.LINE_AA)

cv2.imshow("image",image)

cv2.waitKey(0)
cv2.destroyAllWindows()
```
